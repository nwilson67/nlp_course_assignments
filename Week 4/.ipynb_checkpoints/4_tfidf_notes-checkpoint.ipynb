# 1. TF-IDF

```python
corpus = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first USU document?',
]

vectorizer_default = TfidfVectorizer()
vectorizer_smoothed = TfidfVectorizer(norm=None)
vectorizer_traditional = TfidfVectorizer(norm=None, smooth_idf=False)


X = {
    'default': vectorizer_default.fit_transform(corpus),
    'smoothed': vectorizer_smoothed.fit_transform(corpus),
    'traditional': vectorizer_traditional.fit_transform(corpus)
}

```

```python
vectorizer_traditional.fit_transform(corpus).toarray()
```

```python
vectorizer_traditional.get_feature_names_out()
```

```python
for variation, x in X.items():
    print(variation)
    for i, row in enumerate(x.toarray()):
        zipped_lists = zip(row, vectorizer_default.get_feature_names_out())
        
        # Convert the zip object to a list of tuples for display
        result_list = list(zipped_lists)
        
        print(result_list)
    print(" ")
```

#### By Hand (for the "USU" in the last document, should match above)

```python
# Traditional
tf_usu_4 = 1
idf_usu = np.log(4 / 1) + 1  # 4 documents total (numerator), USU appears in one document (denominator). By adding 1 on the outside we prevent idf from being zero and essentially ignoring a term. Gives all terms a chance to contribute.
print(f"Traditional: {tf_usu_4 * idf_usu}")

# Smoothed
tf_usu_4 = 1
idf_usu = np.log((4 + 1) / (1 + 1)) + 1 # adding 1 above and below provides a smoothing effect by bringing rare words closer to common words. Also avoids out of vocab term undefined errors (0 in the denominator).
print(f"Smoothed: {tf_usu_4 * idf_usu}")

# Normalized
tf_usu_4 = 1
idf_usu = np.log((4 + 1) / (1 + 1)) + 1  
norm = np.sqrt(0.0**2 + 1.2231435513142097**2 + 1.5108256237659907**2 + 1.0**2 + 0.0**2 + 0.0**2 + 1.0**2 + 0.0**2 + 1.0**2 + 1.916290731874155**2)  # norm of the smoothed version
print(f"Normalized: {(tf_usu_4 * idf_usu) / norm} ")
```
